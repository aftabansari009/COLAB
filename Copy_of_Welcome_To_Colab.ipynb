# **making the data frame using dictionary**

````
#sample data
data={
    'Region':['east','west','east','south','west','south'],
    'product_type':['electronics','clothing','electronics','electronics','clothing','clothing'],
    'customres_segment':['regular','premium','regular','premium','regular','premium'],
    'sales':[1000,1500,1200,2000,1800,1300],
    'quantity':[5,10,7,12,8,6]

  }
df=pd.DataFrame(data)
print(df)
#Group By multiple colomns(e.g'Region',Product,Customrs)
grouped_data=df.groupby(['Region','product_type','customres_segment'])

#apply aggregaion functions
aggregated_data=grouped_data['quantity'].sum()
print(aggregated_data)
````
